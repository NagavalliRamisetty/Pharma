# <span style="color: blue;">**Analyzing Molecular Properties Dataset from PubChem**</span>



```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Loading the data
df = pd.read_csv("CSV (1).csv")
```

```python
print("First few rows of the dataset:")
print(df.head())
```

```python
print("\nDataset information:")
print(df.info())
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())
```

The null values are 0 for every coloumn. So, we can proceed with this dataset.

```python
sns.set(style="whitegrid")
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="MolecularWeight", y="XLogP", hue="CID", palette="viridis", s=100)
plt.title("Molecular Weight vs XLogP", fontsize=14)
plt.xlabel("Molecular Weight")
plt.ylabel("XLogP")
plt.legend(title="CID")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="CID", y="XLogP", palette="magma")
plt.title("XLogP Value for Each Compound", fontsize=14)
plt.xlabel("Compound ID (CID)")
plt.ylabel("XLogP")
plt.tight_layout()
plt.show()
```

```python
print("""
Bonus:
Understanding molecular weight and LogP is important in drug development.

- Compounds with higher LogP are more lipophilic (fat-soluble), while those with lower or negative LogP are more hydrophilic (water-soluble).
- Molecular weight influences how easily a compound can be absorbed or transported.

These properties help in selecting suitable solvents and delivery systems for drugs, ensuring they dissolve properly and reach the target site in the body effectively.
""")
```
